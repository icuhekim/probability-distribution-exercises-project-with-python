## Binomial distribution

In probability theory and statistics, the binomial distribution is the discrete probability distribution that gives only two possible results in an experiment; for example, when you flip a coin, it can be either head or tails.

```python
from scipy.stats import binom

# Define all the probability functions related to this distribution

def dbinom(x, size, prob = 0.5):
    """
    Calculates the point estimate of the binomial distribution
    """
    result = binom.pmf(k = x, n = size, p = prob, loc = 0)

    return result

def pbinom(q, size, prob = 0.5):
    """
    Calculates the cumulative of the binomial distribution
    """
    result = binom.cdf(k = q, n = size, p = prob, loc = 0)

    return result

def qbinom(p, size, prob = 0.5):
    """
    Calculates the quantile function from the binomial distribution
    """
    result = binom.ppf(q = p, n = size, p = prob, loc = 0)

    return result

def rbinom(n, size, prob = 0.5):
    """
    Generates random variables from the binomial distribution
    """
    result = binom.rvs(n = size, p = prob, size = n)

    return result
```

```python
# Binomial(10, 0.2) distribution, where n = 10 is the number of trials and p = 0.2 is the probability of success in a single trial
import numpy as np

np.random.seed(42)

print(f"Probability that a Binomial(10, 0.2) takes the value 2: {dbinom(2, size = 10, prob = 0.2)}")
print(f"Probability that a Binomial(10, 0.2) will take a value less than 2: {pbinom(2, size = 10, prob = 0.2)}")
print(f"Which value of a Binomial(10, 0.2) has a cumulative probability of 0.9? {qbinom(0.9, size = 10, prob = 0.2)}")
print(f"Generate 2000 random values from a Binomial(10, 0.2) distribution: {rbinom(2000, size = 10, prob = 0.2)}")
```

```python
import matplotlib.pyplot as plt

np.random.seed(42)

size = 10
prob = 0.2
data = rbinom(2000, size = 10, prob = 0.2)

pmf = dbinom(data, size = size, prob = prob)
cdf = pbinom(data, size = size, prob = prob)
ppf = qbinom(data, size = size, prob = prob)

fig, axis = plt.subplots(3, 1, figsize = (5, 7))

axis[0].stem(data, pmf, "r-")
axis[1].stem(data, cdf, "b-")
axis[2].stem(data, ppf, "g-")

# Adjust the layout
plt.tight_layout()

# Show the plot
plt.show()
```

## Exercise 1

Write the above functions but for both normal and uniform distributions.

```python
from scipy.stats import norm

# Probability density function
def dnorm(x, mean=0, sd=1):
    """
    Calculates the probability density of the normal distribution
    """
    result = norm.pdf(x=x, loc=mean, scale=sd)
    return result


# Cumulative distribution function
def pnorm(q, mean=0, sd=1):
    """
    Calculates the cumulative probability of the normal distribution
    """
    result = norm.cdf(x=q, loc=mean, scale=sd)
    return result


# Quantile function (inverse CDF)
def qnorm(p, mean=0, sd=1):
    """
    Calculates the quantiles of the normal distribution
    """
    result = norm.ppf(q=p, loc=mean, scale=sd)
    return result


# Random samples
def rnorm(n, mean=0, sd=1):
    """
    Generates random variables from the normal distribution
    """
    result = norm.rvs(loc=mean, scale=sd, size=n)
    return result
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

mean = 0
sd = 1

# Generate 2,000 random observations
data = rnorm(2000, mean=mean, sd=sd)

# Values from four standard deviations below to four above the mean
x = np.linspace(mean - 4 * sd, mean + 4 * sd, 500)

# Calculate PDF and CDF
pdf = dnorm(x, mean=mean, sd=sd)
cdf = pnorm(x, mean=mean, sd=sd)

# Probabilities and their corresponding quantiles
p = np.linspace(0.01, 0.99, 500)
ppf = qnorm(p, mean=mean, sd=sd)

# Create three plots
fig, axis = plt.subplots(3, 1, figsize=(6, 10))

# Probability density function
axis[0].plot(x, pdf, "r-", linewidth=2)
axis[0].set_title("Standard Normal Distribution - PDF")
axis[0].set_xlabel("x")
axis[0].set_ylabel("Density")
axis[0].grid(True)

# Cumulative distribution function
axis[1].plot(x, cdf, "b-", linewidth=2)
axis[1].set_title("Standard Normal Distribution - CDF")
axis[1].set_xlabel("x")
axis[1].set_ylabel("Cumulative Probability")
axis[1].grid(True)

# Quantile function
axis[2].plot(p, ppf, "g-", linewidth=2)
axis[2].set_title("Standard Normal Distribution - Quantile Function")
axis[2].set_xlabel("Probability")
axis[2].set_ylabel("Quantile")
axis[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import uniform

# Define all the probability functions related to this distribution

def dunif(x, min=0, max=1):
    """
    Calculates the probability density function (PDF)
    """
    result = uniform.pdf(x=x, loc=min, scale=max-min)
    return result


def punif(q, min=0, max=1):
    """
    Calculates the cumulative distribution function (CDF)
    """
    result = uniform.cdf(x=q, loc=min, scale=max-min)
    return result


def qunif(p, min=0, max=1):
    """
    Calculates the quantile function
    """
    result = uniform.ppf(q=p, loc=min, scale=max-min)
    return result


def runif(n, min=0, max=1):
    """
    Generates random variables from the uniform distribution
    """
    result = uniform.rvs(loc=min, scale=max-min, size=n)
    return result
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

minimum = 0
maximum = 1

# Generate random observations
data = runif(2000, min=minimum, max=maximum)

# x-values for plotting
x = np.linspace(minimum, maximum, 500)

# Calculate PDF and CDF
pdf = dunif(x, min=minimum, max=maximum)
cdf = punif(x, min=minimum, max=maximum)

# Probabilities for the quantile function
p = np.linspace(0.01, 0.99, 500)
ppf = qunif(p, min=minimum, max=maximum)

# Create plots
fig, axis = plt.subplots(3, 1, figsize=(6, 10))

# PDF
axis[0].plot(x, pdf, "r-", linewidth=2)
axis[0].set_title("Uniform Distribution - PDF")
axis[0].set_xlabel("x")
axis[0].set_ylabel("Density")
axis[0].grid(True)

# CDF
axis[1].plot(x, cdf, "b-", linewidth=2)
axis[1].set_title("Uniform Distribution - CDF")
axis[1].set_xlabel("x")
axis[1].set_ylabel("Cumulative Probability")
axis[1].grid(True)

# Quantile Function
axis[2].plot(p, ppf, "g-", linewidth=2)
axis[2].set_title("Uniform Distribution - Quantile Function")
axis[2].set_xlabel("Probability")
axis[2].set_ylabel("Quantile")
axis[2].grid(True)

# Adjust layout
plt.tight_layout()

# Show the plots
plt.show()
```
